```python
import requests, numpy as np, pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

print("Trần Chí Nguyên - 6451071052")
FILE_ID = "1XSKNTP_JSRE2Bt-g_Spx3sqftK-3dif_"  # bank-full.csv trong thư mục dataset của môn học
def load_bank():
    path = Path("/content/bank-full.csv")
    if not path.exists():
        response = requests.get("https://drive.google.com/uc", params={"export": "download", "id": FILE_ID}, timeout=120)
        response.raise_for_status()
        path.write_bytes(response.content)
    return pd.read_csv(path, sep=";")
df = load_bank()
print("Kích thước:", df.shape)
display(df.head())
```

## Thực hiện yêu cầu

Cell dưới đây chứa lời giải cho câu này.

```python
numeric = df.select_dtypes(include="number")
display(numeric.describe().T)
if not numeric.empty:
    plt.figure(figsize=(10, 7)); sns.heatmap(numeric.corr(), annot=True, cmap="coolwarm", center=0, fmt=".2f")
    plt.title("Tương quan giữa các thuộc tính số"); plt.tight_layout(); plt.show()
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(data=df, x="age", hue="y", bins=30, element="step", stat="density", common_norm=False, ax=axes[0])
sns.boxplot(data=df, x="y", y="duration", ax=axes[1])
axes[0].set_title("Tuổi theo kết quả đăng ký"); axes[1].set_title("Thời lượng contact theo kết quả")
plt.tight_layout(); plt.show()
print("Lưu ý: duration chỉ biết sau cuộc gọi, có thể gây rò rỉ thông tin nếu dùng để dự đoán trước khi gọi.")
```
