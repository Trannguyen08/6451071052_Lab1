```python
import requests, numpy as np, pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

print("Trần Chí Nguyên - 6451071052")
FILE_ID = "1XSKNTP_JSRE2Bt-g_Spx3sqftK-3dif_"  # bank-full.csv trong thư mục dataset của môn học
def load_bank():
    path = Path("/content/bank-full.csv")
    if not path.exists():
        response = requests.get("https://drive.google.com/uc", params={"export": "download", "id": FILE_ID}, timeout=120)
        response.raise_for_status()
        path.write_bytes(response.content)
    return pd.read_csv(path, sep=";")
df = load_bank()
print("Kích thước:", df.shape)
display(df.head())
```

## Thực hiện yêu cầu

Cell dưới đây chứa lời giải cho câu này.

```python
work = df.copy()
for col in work.select_dtypes(include="number"):
    work[col] = work[col].fillna(work[col].median())
for col in work.select_dtypes(exclude="number"):
    mode = work[col].mode(dropna=True)
    work[col] = work[col].fillna(mode.iat[0] if not mode.empty else "unknown")
print("Số ô NaN sau điền:", int(work.isna().sum().sum()))
print("Số giá trị unknown theo cột:")
display((work.astype("string").apply(lambda s: s.str.lower().eq("unknown").sum())).sort_values(ascending=False).to_frame("unknown_count").head(10))
print("Các nhãn phân loại được giữ nguyên để tránh tự ý gán unknown thành một giá trị đã biết.")
```
