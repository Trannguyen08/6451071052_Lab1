```python
import requests, numpy as np, pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

print("Trần Chí Nguyên - 6451071052")
FILE_ID = "1TCnJqaowkv0caJTJrmjoqb0AB2bMvgxP"
def load_titanic():
    path = Path("/content/titanic.csv")
    if not path.exists():
        response = requests.get("https://drive.google.com/uc", params={"export": "download", "id": FILE_ID}, timeout=90)
        response.raise_for_status()
        path.write_bytes(response.content)
    data = pd.read_csv(path)
    data.columns = [str(c).strip().lower().replace("_", ".") for c in data.columns]
    return data
df = load_titanic()
print("Kích thước dữ liệu:", df.shape)
display(df.head())
```

## Thực hiện yêu cầu

Cell dưới đây chứa lời giải cho câu này.

```python
family = df.copy()
family["family_count"] = family["sibsp"].fillna(0) + family["parch"].fillna(0)
family["family_group"] = pd.cut(family["family_count"], bins=[-1, 0, 3, 6, np.inf], labels=["Đi một mình", "1–3", "4–6", "7+"])
display(family.groupby("family_group", observed=False).agg(passengers=("survived", "size"), survival_rate=("survived", "mean")).round(3))
plt.figure(figsize=(8, 5))
sns.barplot(data=family, x="family_group", y="survived", errorbar=None)
plt.title("Tỷ lệ sống theo số người thân đi cùng"); plt.xlabel("Số người thân đi cùng"); plt.ylabel("Tỷ lệ sống sót"); plt.ylim(0, 1); plt.tight_layout(); plt.show()
print("Nhận xét: đi một mình và đi cùng nhóm nhỏ có kết quả khác với nhóm gia đình đông; nhóm rất đông thường có tỷ lệ sống thấp hơn.")
```
