```python
import requests, numpy as np, pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

print("Trần Chí Nguyên - 6451071052")
FILE_ID = "1TCnJqaowkv0caJTJrmjoqb0AB2bMvgxP"
def load_titanic():
    path = Path("/content/titanic.csv")
    if not path.exists():
        response = requests.get("https://drive.google.com/uc", params={"export": "download", "id": FILE_ID}, timeout=90)
        response.raise_for_status()
        path.write_bytes(response.content)
    data = pd.read_csv(path)
    data.columns = [str(c).strip().lower().replace("_", ".") for c in data.columns]
    return data
df = load_titanic()
print("Kích thước dữ liệu:", df.shape)
display(df.head())
```

## Thực hiện yêu cầu

Cell dưới đây chứa lời giải cho câu này.

```python
plot_data = df.dropna(subset=["embarked", "pclass", "survived", "sex"]).copy()
fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=True)
for ax, gender in zip(axes, ["female", "male"]):
    sub = plot_data[plot_data["sex"].str.lower() == gender]
    sns.barplot(data=sub, x="embarked", y="survived", hue="pclass", errorbar=None, ax=ax)
    ax.set(title=f"Giới tính: {gender}", xlabel="Cổng lên tàu", ylabel="Tỷ lệ sống sót")
    ax.set_ylim(0, 1)
plt.suptitle("Tỷ lệ sống theo cổng, hạng vé và giới tính"); plt.tight_layout(); plt.show()
display(plot_data.groupby(["sex", "embarked", "pclass"], observed=True)["survived"].agg(["mean", "count"]).round(3))
print("Nhận xét: hạng vé và giới tính thể hiện khác biệt rõ về tỷ lệ sống; cổng lên tàu có thể liên quan nhưng cần diễn giải cùng cơ cấu hành khách từng nhóm.")
```
