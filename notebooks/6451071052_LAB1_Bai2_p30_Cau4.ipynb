```python
import requests, pandas as pd, numpy as np, matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

print("Trần Chí Nguyên - 6451071052")
path = Path("/content/tb_tracking.xlsx")
if not path.exists():
    FILE_ID = "1Xizh3zvtMF4IdRNlueb74meOXk80-CHc"
    response = requests.get("https://drive.google.com/uc", params={"export": "download", "id": FILE_ID}, timeout=90)
    response.raise_for_status()
    path.write_bytes(response.content)
df = pd.read_excel(path, header=None)
columns = ["id", "device_id", "co_level", "humidity", "temperature", "time"]
if df.shape[1] != len(columns):
    raise ValueError(f"Cần 6 cột dữ liệu, nhận được {df.shape[1]}.")
df.columns = columns
print("Kích thước:", df.shape)
display(df.head())
```

## Thực hiện yêu cầu

Cell dưới đây chứa lời giải cho câu này.

```python
work = df.copy(); work["time"] = pd.to_datetime(work["time"], errors="coerce", dayfirst=True)
work = work.dropna(subset=["time", "co_level", "humidity", "temperature"])
work = work.loc[work["time"] >= pd.Timestamp("2018-01-03")].copy()
work["minutes"] = work["time"].dt.hour * 60 + work["time"].dt.minute
plt.figure(figsize=(10, 5)); plt.scatter(work["minutes"], work["co_level"], alpha=.45, s=12)
plt.title("Nồng độ CO theo phút trong ngày"); plt.xlabel("Phút trong ngày"); plt.ylabel("CO level"); plt.tight_layout(); plt.show()
print("Tương quan minutes–CO:", round(work["minutes"].corr(work["co_level"]), 3))
```
