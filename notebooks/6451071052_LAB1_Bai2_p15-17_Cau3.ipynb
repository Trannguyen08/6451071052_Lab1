```python
import requests, numpy as np, pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

print("Trần Chí Nguyên - 6451071052")
FILE_ID = "1TCnJqaowkv0caJTJrmjoqb0AB2bMvgxP"
def load_titanic():
    path = Path("/content/titanic.csv")
    if not path.exists():
        response = requests.get("https://drive.google.com/uc", params={"export": "download", "id": FILE_ID}, timeout=90)
        response.raise_for_status()
        path.write_bytes(response.content)
    data = pd.read_csv(path)
    data.columns = [str(c).strip().lower().replace("_", ".") for c in data.columns]
    return data
df = load_titanic()
print("Kích thước dữ liệu:", df.shape)
display(df.head())
```

## Thực hiện yêu cầu

Cell dưới đây chứa lời giải cho câu này.

```python
plot_data = df.dropna(subset=["age", "sex", "survived"]).copy()
plot_data["age_group"] = pd.cut(plot_data["age"], bins=[0, 10, 18, 22, 27, 33, 40, 66, np.inf], right=True,
                                labels=["≤10", "11–18", "19–22", "23–27", "28–33", "34–40", "41–66", ">66"])
fig, ax = plt.subplots(figsize=(11, 5))
sns.barplot(data=plot_data, x="age_group", y="survived", hue="sex", errorbar=None, ax=ax)
ax.set(title="Tỷ lệ sống theo nhóm tuổi và giới tính", xlabel="Nhóm tuổi", ylabel="Tỷ lệ sống sót")
ax.set_ylim(0, 1); plt.tight_layout(); plt.show()
print("Nhận xét: tỷ lệ sống sót thường cao hơn ở nữ; mức sống sót khác nhau giữa các nhóm tuổi. Nhóm tuổi lớn có ít quan sát hơn nên cần xem cùng số lượng mẫu.")
```
