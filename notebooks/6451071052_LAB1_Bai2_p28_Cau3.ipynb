```python
import requests, numpy as np, pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

print("Trần Chí Nguyên - 6451071052")
FILE_ID = "1XSKNTP_JSRE2Bt-g_Spx3sqftK-3dif_"  # bank-full.csv trong thư mục dataset của môn học
def load_bank():
    path = Path("/content/bank-full.csv")
    if not path.exists():
        response = requests.get("https://drive.google.com/uc", params={"export": "download", "id": FILE_ID}, timeout=120)
        response.raise_for_status()
        path.write_bytes(response.content)
    return pd.read_csv(path, sep=";")
df = load_bank()
print("Kích thước:", df.shape)
display(df.head())
```

## Thực hiện yêu cầu

Cell dưới đây chứa lời giải cho câu này.

```python
print("Tỷ lệ đăng ký term deposit:")
display(df["y"].value_counts(dropna=False).to_frame("count").assign(percent=lambda x: (x["count"] / len(df) * 100).round(2)))
target = df.groupby("job", dropna=False)["y"].apply(lambda s: (s == "yes").mean()).sort_values(ascending=False)
plt.figure(figsize=(11, 5)); target.plot(kind="bar", color="teal")
plt.title("Tỷ lệ đăng ký theo nghề nghiệp"); plt.ylabel("Tỷ lệ đăng ký"); plt.xlabel("Nghề nghiệp"); plt.tight_layout(); plt.show()
display(target.rename("subscription_rate").to_frame().round(3))
```
