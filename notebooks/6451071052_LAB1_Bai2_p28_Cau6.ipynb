```python
import requests, numpy as np, pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

print("Trần Chí Nguyên - 6451071052")
FILE_ID = "1XSKNTP_JSRE2Bt-g_Spx3sqftK-3dif_"  # bank-full.csv trong thư mục dataset của môn học
def load_bank():
    path = Path("/content/bank-full.csv")
    if not path.exists():
        response = requests.get("https://drive.google.com/uc", params={"export": "download", "id": FILE_ID}, timeout=120)
        response.raise_for_status()
        path.write_bytes(response.content)
    return pd.read_csv(path, sep=";")
df = load_bank()
print("Kích thước:", df.shape)
display(df.head())
```

## Thực hiện yêu cầu

Cell dưới đây chứa lời giải cho câu này.

```python
work = df.copy()
work["y_binary"] = work["y"].map({"no": 0, "yes": 1})
work = work.drop(columns=["y"])
categorical = work.select_dtypes(include=["object", "string"]).columns.tolist()
processed = pd.get_dummies(work, columns=categorical, dummy_na=False, dtype=int)
print("Dữ liệu sau one-hot encoding:", processed.shape)
display(processed.head())
print("Số giá trị rỗng:", int(processed.isna().sum().sum()))
print("Tỷ lệ nhãn yes:", round(work["y_binary"].mean(), 4))
```
